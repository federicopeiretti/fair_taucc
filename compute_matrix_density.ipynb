```python
import numpy as np
from scipy import sparse
import os
```

# Synthetic matrices

```python
dataset = "synthetic"

for clus in [3,5,10,"3_rc"]:
    root = os.getcwd() + f"/datasets/{dataset}/clus{clus}"
    matrix = np.load(root + "/matrix.npy")
    values, counts = np.unique(matrix, return_counts=True)
    density = np.sum(counts[1:])/np.sum(counts)
    print(f"clus{clus} - matrix density: {density:.4f}")
```

# Real-world datasets 

```python
datasets = ["movielens-1m", "yelp", "amazon", "lfw"]

for ds in datasets:
    
    if "movielens" in ds:
        root = os.getcwd() + f"/datasets/movielens/{ds}"
    else:
        root = os.getcwd() + f"/datasets/{ds}"
    
    matrix = np.load(root + "/matrix.npy")
    values, counts = np.unique(matrix, return_counts=True)
    density = np.sum(counts[1:])/np.sum(counts)
    print(f"{ds} - matrix density: {density:.4f}")
```
