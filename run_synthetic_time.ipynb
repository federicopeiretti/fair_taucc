```python
import time
import os
import numpy as np
import pandas as pd
from tauCC.src.fairness_metrics import balance_gen
```

```python
algorithm = "taucc_fair_max"

if algorithm == "taucc_fair":
    from tauCC.src.taucc.taucc_fair_rows_cols import FairCoclus
elif algorithm == "taucc_fair_max":
    from tauCC.src.taucc.taucc_fair_rows_cols_max import FairCoclus

INIT = "random"
RUNS = 10

NUM_CLUSTERS = 10
NUM_GROUPS = 2
sensitive_degree = 1.0

root = os.getcwd()
#print(f"root: {root}")

PATH_DATASET = f"{root}/datasets/synthetic/clus{NUM_CLUSTERS}"
PATH_SENSITIVE = f"{PATH_DATASET}/groups{NUM_GROUPS}"

V = np.load(PATH_DATASET + f"/matrix.npy", allow_pickle=True)
true_rows = np.load(PATH_DATASET + f"/row_labels.npy", allow_pickle=True)
true_cols = np.load(PATH_DATASET + f"/col_labels.npy", allow_pickle=True)

Sx = np.load(PATH_SENSITIVE + f"/sensitive_p{sensitive_degree}.npy")

print(f"algorithm: {algorithm}")
print(PATH_DATASET)
print(PATH_SENSITIVE)
```

```python
groups_value, groups_count = np.unique(Sx, return_counts=True)
groups_count
```

```python
topsis = pd.read_csv(f"/home/peiretti/fair-clustering/results/synthetic/clus{NUM_CLUSTERS}/{algorithm}/topsis_best_param_groups{NUM_GROUPS}.csv")
df = topsis.query("sensitive_p==@sensitive_degree")
df
```

```python
fair_params = np.array([0.5,0.4])
times = []
balance = []
```

```python
for run in np.arange(0, RUNS):
    
    start_time = 0
    end_time = 0

    start_time = time.time()

    if NUM_CLUSTERS < 10:
        model = FairCoclus(initialization=INIT, verbose=True, k=10, l=10)
    else:
        model = FairCoclus(initialization=INIT, verbose=True, k=20, l=20)

    model.fit(V=V, Sx=Sx, fair_row_parameters=fair_params)
    end_time = time.time()
    execution_time = end_time - start_time
    times.append(execution_time)
    
    predict_rows = model.row_labels_
    balance_bera = balance_gen(Sx, predict_rows)
    balance.append(balance_bera)
```

```python
time_mean = np.round(np.mean(times),2)
time_std = np.round(np.std(times),2)
balance_mean = np.mean(balance)
balance_std = np.std(balance)
```
