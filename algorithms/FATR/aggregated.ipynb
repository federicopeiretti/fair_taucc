```python
from unittest import result

import pandas as pd

def get_aggregated(df, columns, exclude):
    metrics = [c for c in df.select_dtypes(include='number').columns if c not in exclude]
    agg_funcs = ['mean', 'std', 'var']
    agg = df[metrics].agg(['mean', 'std', 'var'])
    result = {}
    for col in metrics:
        result[f'{col}_mean'] = agg.loc['mean', col]
        result[f'{col}_std']  = agg.loc['std',  col]
        result[f'{col}_var']  = agg.loc['var',  col]
    result_df = pd.DataFrame([result])
    return result_df
```

```python
dataset = ["amazon", "yelp", "lfw", "movielens-1m", "movielens-1m"]
sensitive = ["gender", "gender", "gender", "gender", "age"]

for idx, dataset in enumerate(dataset):

    if dataset == "movielens-1m":
        path = f"results/{dataset}/{sensitive[idx]}/"
    else: 
        path = f"results/{dataset}/"

    df_fair = pd.read_csv(path + "fair/results_runs.csv")
    df_vanilla = pd.read_csv(path + "vanilla/results_runs.csv")

    exclude = {'run', 'row_clus', 'col_clus'}
    columns = ['run']
    
    print("Dataset:", dataset)
    print("Sensitive attribute:", sensitive[idx])

    get_aggregated(df_fair, columns, exclude).to_csv(path + "fair/aggregated.csv", index=False)
    get_aggregated(df_vanilla, columns, exclude).to_csv(path + "vanilla/aggregated.csv", index=False)
```
